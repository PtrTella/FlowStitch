### 9.1 Dai 2D ai 4D: Oggetti come Tubi Topologici
L'estensione della Chirurgia Latente Topologica dal dominio immagine (2D) al dominio video (3D spaziotemporale + canali) non richiede modifiche strutturali alla teoria del Flow Matching (*Lipman et al., 2023*), ma solo un incremento dimensionale del manifold.

Se in un'immagine 2D un oggetto isolato corrisponde a un generatore $H_0$ (componente connessa ad alta coesione di velocità), in un tensore video di coordinate $(t_{video}, x, y)$ lo stesso oggetto forma una struttura continua detta **Tubo Topologico**. 
In termini di omologia persistente, le sezioni trasversali si legano nel tempo creando un invariante omologico $H_1$ (se trattiamo i limiti del tubo) o un voluminoso $H_0$ nel cubo spaziotemporale.
