Questo notebook funge da base per le fasi finali e più avanzate del progetto **FlowStitch**, in cui i "cartamodelli" estratti verranno utilizzati in un ecosistema di caching vettoriale.

### Fase 3: Latent Stitching (Immediato)
Sviluppare l'algoritmo di composizione tensoriale. Dato un tensore sorgente $A$ e un tensore target $B$, l'integrazione avverrà tramite la maschera $M$:
$$x_{stitch} = M \odot x_B + (1 - M) \odot x_A$$
Questa operazione deve essere validata per verificare se il modello riesce a "cucire" i bordi dell'oggetto nel nuovo contesto senza artefatti visivi.

### Fase 4: Integrazione FAISS (Database Vettoriale)
Implementare un sistema di caching dove i vettori $x_{pred}$ vengono indicizzati. 
* **Obiettivo:** Se un nuovo prompt ha una struttura geometrica simile a uno già processato (calcolata tramite *Cosine Similarity*), il sistema deve recuperare il latente pre-calcolato invece di rigenerarlo, riducendo drasticamente il tempo di inferenza.

### Fase 5: Validazione su Prompt Complessi
