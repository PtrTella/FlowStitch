### Formulazione Hamiltoniana del Flow Matching

$$ \frac{dq}{dt} = \frac{\partial \mathcal{H}}{\partial p}, \quad \frac{dp}{dt} = -\frac{\partial \mathcal{H}}{\partial q} $$
